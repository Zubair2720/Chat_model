```python
import os
from google import genai
from google.genai import types

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY"),
    http_options=types.HttpOptions(timeout=60000)  # 60 sec
)

for model in ["gemini-2.5-flash-lite", "gemini-2.5-flash", "gemini-3.5-flash-lite"]:
    try:
        r = client.models.generate_content(model=model, contents="Explain Generative AI?")
        print(f"({model} se jawab aaya)\n")
        print(r.text)
        break
    except Exception as e:
        print(f"{model} fail: {e}\n")
```

```python

import time
from google import genai

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)

chat = client.chats.create(
    model="gemini-3.5-flash-lite"
)

while True:
    prompt = input("User: ")

    if prompt.lower() in ['bye', 'exit', 'stop']:
        print("Stopping...")
        time.sleep(2)
        print("Bye-Bye!")
        break

    response = chat.send_message(message=prompt)

    print(f"Gemini: {response.text}")

    print('\n' + '*' * 100 + '\n')
```
